# Data Preparation and Analysis

```python
import pandas as pd
```

### Dummy data

```python
student_data = {'id' : ['1001-2', '1002-2', '1003-2', '1004-2', '1005-2', '1006-2'],
               'name' : ['Louisa Decker', 'Sioned Allan', 'Cayden Collier', 'Jerome Rudd', 'Jibril Morrow', 'David Strouut'],
                'section' : ['1A', '2A', '3B', '3D', '2C', '1A'],
                'year' : ['BSIT-1', 'BSIT-2', 'BSIT-3', 'BSIT-3', 'BSIT-2', 'BSIT-1']
               }
```

#### Student Data Frame from 'student_data' dictionary

```python
student_data_frame = pd.DataFrame(student_data)
```

```python
grades_data = {'id' : ['1001-2', '1002-2', '1003-2', '1004-2', '1005-2', '1006-2'],
              'gpa' : [91., 92., 88., 83., 75., 80.]
              }
```

#### Grades Data Frame from 'grades_data'

```python
grades_data_frame = pd.DataFrame(grades_data)
```

#### Merged student and grade Data Frame

```python
student_df = pd.merge(student_data_frame, grades_data_frame, how='inner', on='id')
```

```python
student_df
```

```python
student_df.info()
```

#### deans_list function

```python
def deans_list(gpa):
    return 'Y' if gpa >=90.0 else 'N'
```

```python
student_df['deans_list'] = student_df['gpa'].apply(deans_list)
```

```python
student_df
```

#### Student who are not on deans_list

```python
student_df[student_df['deans_list'] == 'N'][['name', 'year', 'section']]
```

#### First year students who made to the deans_list

```python
student_df[(student_df['year'] == 'BSIT-1') & (student_df['deans_list'] == 'Y')][['name', 'year', 'section']]
```

#### Deans list with a GPA of 90 or higher

```python
student_df[(student_df['gpa'] >= 90.0) & (student_df['deans_list'] == 'Y')][['name', 'year', 'section']]
```

#### Average GPA of student per year level

```python
avg_gpa_per_year = student_df.groupby('year')['gpa'].mean().reset_index()
```

```python
avg_gpa_per_year_col = ['year', 'gpa']
```

```python
avg_gpa_per_year
```

#### Year level with the highest GPA

```python
avg_gpa_per_year.loc[avg_gpa_per_year['gpa'].idxmax(), 'year']
```

#### Average GPA of BSIT-3 students

```python
avg_gpa_per_year.loc[avg_gpa_per_year['year'] == 'BSIT-3', 'gpa']
```

#### Highest GPA per year level

```python
student_df.loc[student_df.groupby('year')['gpa'].idxmax(), ['year', 'id', 'name', 'section', 'gpa', 'deans_list']]
```
